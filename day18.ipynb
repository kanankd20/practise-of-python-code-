```python
import pandas as pd 
import numpy as np 
```

```python
data={
    "dept": ["it","it","it","it","finance","hr"],
    "salary":[10000,20000,30000,40000,50000,60000],
    "employee":[1,2,3,4,5,6],
    "experience":[1,2,3,4,5,6]
}
```

```python
df1 = pd.DataFrame(data)
df1
```

```python
avg_salary=df1.groupby("dept")["salary"].mean()
print(avg_salary)
```

```python
max_experience=df1.groupby("dept")["experience"].max()
print(max_experience)
```

```python
total_salary=df1.groupby("dept")["salary"].sum()
total_salary
```

```python
employee_count = df1.groupby("dept")["employee"].count()
employee_count
```

```python
average_salary_experience = df1.groupby("dept")[["salary","experience"]].mean()
average_salary_experience
```

```python
df1.groupby("dept").agg({"salary":["mean","sum"],"experience":["mean","max"],"employee":"count"})
```

```python
employee =pd.DataFrame({
    "name":["a","b","c","d"],
    "emp_id":[1,2,3,4]
})
```

```python
departments=pd.DataFrame({
    "emp_id":[1,2,3,4],
    "dept":["it","hr","finance","it"]
})
```

```python
inner_merge = pd.merge(employee, departments, left_on='email_id', right_on='emp_id', how='inner')
inner_merge
```

```python
df1= pd.DataFrame({
    "name":["a","b","c","d"],
    "email_id":[1,2,3,4]
})
df2=pd.DataFrame({
    "name":["e","f","g","h"],
    "email_id":[5,6,7,8]
})
row_concat = pd.concat([df1,df2])
row_concat
```

```python
df=pd.concat([df1,df2],ignore_index=True)
df
```

```python
column_concatenate= pd.concat([df1,df2],axis=1,ignore_index=True)
column_concatenate
```

```python
concate_keys=pd.concat([df1,df2],keys=["df1","df2"])
concate_keys
```

```python
p=pd.pivot_table(df,index="name",values="email_id",aggfunc="sum")
p
```

```python
products=pd.DataFrame({
    "product_id":[1,2,3,4],
    "product_name":["a","b","c","d"],
    "price":[100,200,300,400],
    "product_sold":[10,20,30,40]
})

sales=pd.DataFrame({
    "product_id":[1,2,3,4],
    "quantity_sold":[10,20,30,40]
})
result=pd.merge(products,sales)
result
```

```python
total_sales=pd.pivot_table(result,index="product_name",values="quantity_sold",aggfunc="sum")
total_sales
```
