```python
import pandas as pd
import numpy as np 
```

```python
data={
    "department":["it","it","it","it","finance","hr"],
    "salary":[10000,20000,30000,40000,50000,60000],
    "employee":[1,2,3,4,5,6],
    "experience":[1,2,3,4,5,6]
}
df=pd.DataFrame(data)
df
```

```python
#groupby
avg_salary=df.groupby("department")["salary"].mean()
avg_salary
```

```python
max_salary=df.groupby("department")["salary"].max()
max_salary
```

```python
total_salary=df.groupby("department")["salary"].sum()
total_salary
```

```python
emp_count= df.groupby("department")["employee"].count()
emp_count
```

```python
avg_salary_experience= df.groupby("department")[["salary","experience"]].mean()
avg_salary_experience
```

```python
#merge &join
employee =pd.DataFrame({
    "id":[1,2,3,4],
    "name":["a","b","c","d"],
    "department":["it","it","hr","hr"]
})
departemnts=pd.DataFrame({
    "id":[1,2,3,4],
    "salary":[10000,20000,30000,40000],
    "department":["it","hr","finance","it"]
})
result=pd.merge(employee,departemnts,on="id")
result
```

```python
inner_join =pd.merge(employee,departemnts,how="inner",on="id")
inner_join
```

```python
left_join=pd.merge(employee,departemnts,how="left",on="id")
left_join
```

```python
right_join=pd.merge(employee,departemnts,how="right",on="id")
right_join
```

```python
#find records exist in one only 
merged=pd.merge(employee,departemnts,how="outer",indicator=True)
merged
```

```python
#concatenation 
df1=pd.DataFrame({
    "id":[1,2,3,4],
    "name":["a","b","c","d"],
    "salary":[10000,20000,30000,40000],
    "department":["it","it","hr","hr"]
})
df2=pd.DataFrame({
    "id":[1,2,3,4],
    "name":["e","f","g","h"],
    "salary":[50000,60000,70000,80000],
    "department":["it","it","hr","hr"]
})
row_concat=pd.concat([df1,df2])
row_concat
```

```python
#pivot table
data={
    "sales":[10,20,30,40],
    "price":[100,200,300,400],
    "product_id":[1,2,3,4],
    "product_name":["a","b","c","d"],
    "region":["north","south","east","west"]
}
df= pd.DataFrame(data)
df
```

```python
#create pivot table showing total sales by region 
pivot_table=df.pivot_table(index="region",values="sales",aggfunc="sum")
pivot_table
```

```python
#pivot tables with regions as rows and products as columns
pivot_table=df.pivot_table(index="region",columns="product_name",values="sales",aggfunc="sum")
pivot_table
```

```python
#pivot table showing average sales for each product in each region 
pivot1=pd.pivot_table(df,index=["region","product_name"],values="sales",aggfunc="mean")
pivot1
```

```python
#resample daily sales data into weekly sales totals 
sales=pd.DataFrame({
    "date":["2023-01-01","2023-01-02","2023-01-03","2023-01-04","2023-01-05","2023-01-06","2023-01-07","2023-01-08","2023-01-09"],
    "sales":[100,200,300,400,500,600,700,800,900]
})
sales["date"]=pd.to_datetime(sales["date"])
sales=sales.set_index("date")
sales=sales.resample("W").sum()
sales
```

```python
#cumulative sales over time 
sales=pd.DataFrame({
    'dates':['2023-01-01','2023-01-02','2023-01-03','2023-01-04','2023-01-05','2023-01-06','2023-01-07','2023-01-08','2023-01-09'],
    'sales':[100,200,300,400,500,600,700,800,900]
})
sales['dates']=pd.to_datetime(sales['dates'])
sales=sales.set_index('dates')
sales=sales.cumsum()
sales
```
