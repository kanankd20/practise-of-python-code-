```python
import numpy as np
```

```python
a1=np.array([10,20,30,40,50])

a2=np.array([60,70,80,90,10])
a2
```

```python
print(a1+a2)
print(a1-a2)
print(a1*a2)
print(a1/a2)
print(a1//a2)
print(a1**a2)
```

```python
arr*100
```

```python
arr1=np.arange(1,31).reshape(6,5)
arr1
```

```python
#deep copy &shallow copy 
a= np.array([10,20,30,40,50])
a
```

```python
b=a
print(b)
```

```python
b=a.view()
b
```

```python
b=a.view()
b[0]=99
print(b,"shallow copy")
print(a,"original content")
```

```python
arr1=np.array([1,2,3,4,5])
arr1
```

```python
arr2=arr1.copy() #deep copy 
arr2[0]=99
```

```python
print(arr2,"deep copy")
print(arr1,"original content")
```
