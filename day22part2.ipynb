```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df=pd.read_csv("attendance_analysis_dataset.csv")
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
sns.histplot(df["Age"],bins=5, kde=False)
plt.title("AGE DISTRIBUTION")
plt.show()
```

```python
sns.scatterplot(x="Study_Hours", y="Final_Score", data=df, color="green")
plt.title("STUDY HOURS VS FINAL SCORE")
plt.show()
```

```python
sns.boxplot(x="Gender",y="Assignment_Score",data=df)
plt.title("Gender vs Assignment Score")
plt.show()
```

```python
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
sns.barplot(x="Participation",y="Final_Score",data=df)
plt.title("Participation vs Final Score")
plt.show()
```

```python
sns.lineplot(x="Extracurricular",y="Final_Score",data=df)
plt.title("Extracurricular vs Final Score")     
plt.show()
```

```python
sns.pairplot(df)
plt.show()
```
