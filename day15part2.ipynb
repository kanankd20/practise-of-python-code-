```python
import numpy as np

A = np.array([[1, 2],[3, 4]])

B = np.array([[5, 6],[7, 8]])

print(A + B)
```

```python
#broadcasting questions 
C=np.array([10,20]) #concept of automatic shape expansion
print(A+C)
```

```python
#operation between array &scaalr 
print(B*2)
#operation between different shaped arrays 
print(B+C)
```

```python
a=np.array([[1,2,3],[4,5,6]])
n=np.array([10,20,30])
print(a+n) #row wise 
b=np.array([[10],[20]])
print(a+b) #column wise 
```

```python
d=np.dot(a,n)
d
```

```python
c=b.view()
print(c,"shallow copy")
print(b,"original data ")
```

```python
e=c.copy()
e[0]=45
print(e,"deep copy")
print(c,"original data")
```

```python

arr = np.array([1, 2, 3, 4, 5])
sliced = arr[1:4]   # slicing
print(arr)
print(sliced)      

sliced[0] = 99
print(arr)          # [1 99 3 4 5]  ← original भी बदल गया
```

```python
#matrix multiplication 
a=np.array([[1,2,3],[4,5,6]])
b=np.array([11,20,30])
c=np.dot(a,b)
print(c)
```

```python
print(a*b,"element wise")
print(c,"matrix multiplication ")
```

```python
print(a.T,"transpose") #row column interchange 
```

```python
#linear algebra operations 
a=np.array([[1,2,3],[4,5,6]])
b=np.array([11,20,30])
c=np.linalg.vecdot(a,b)
print(c)
```

```python
#array stacking 
#vertical stacking 
a=np.array([[1,2,3],[4,5,6]])
b=np.array([11,20,30])
c=np.vstack((a,b))
print(c)
```

```python
#horizontal stacking 
a=np.array([[1,2,3],[4,5,6]])
b=np.array([[11,20,30],[13,14,15]])
c=np.hstack((a,b))
print(c)
```

```python
d=np.column_stack((a,b)) #column stacking 
d
```

```python
e=np.dstack((a,b))#depth or dimension alignmnet stacking
e
```

```python
#array splitting 
a=np.array([[1,2,3],[4,5,6]])
b=np.array([[11,20,30],[13,14,15]]) #row or horizontal splitting 
c=np.hsplit(a,3)
c
```

```python
d=np.vsplit(b,2) #column or vertical splitting 
d
```

```python
e=np.split(a,2) #dividing into equal parts
e
```
