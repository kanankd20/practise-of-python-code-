```python
import matplotlib.pyplot as plt
```

```python
#project
months=["jan","feb","mar","apr","may","june"]
attendance=[97,90,89,88,67,85]
subjects=["math","physics","english","chemistry"]
marks=[70,80,80,89]

study_hours=[1,2,3,4,4]
marks_scatter=[50,56,67,87,90]

activities=["study","play","sleep","social media"]  
activity_time=[7,2,5,1]
        
```

```python
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.plot(months,attendance,marker='o')
plt.title("monthly attendance")

plt.subplot(2,2,2)
plt.bar(subjects,marks,color='green')
plt.title("marks")

plt.subplot(2,2,3)
plt.scatter(study_hours,marks_scatter,color='pink',s=100)
plt.title("marks vs study hours")

plt.subplot(2,2,4)
plt.pie(activity_time, labels=activities, autopct="%1.1f%%")
plt.title("activity time")

plt.tight_layout()
plt.show()
```

```python
# movies project
movies=["pathan","jawaan","dangal","3idiots"]
box_office_in_cr=[1050,1100,1200,1300]
ratings=[7.5,7.6,7.7,7.8]
years=[2018,2019,2020,2021]
budget=[120,110,100,90]
genre=["action","action","sports","drama"]
genre_count=[2,2,1,1]
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.plot(movies,box_office_in_cr,marker='o')
plt.title("box office collection")

plt.subplot(2,2,2)
plt.bar(movies,ratings,color='green')
plt.title("ratings")

plt.subplot(2,2,3)
plt.scatter(years,budget,color='pink',s=100)
plt.title("budget vs years")

plt.subplot(2,2,4)
plt.pie(genre_count, labels=genre, autopct="%1.1f%%")
plt.title("genre count")

plt.tight_layout()
plt.show()
```
