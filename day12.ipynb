```python
import numpy as np
```

```python
arr=np.arange(1,31).reshape(6,5)
print(arr)
```

```python
print(arr[0])#extract 1st row
print(arr[-1])#extract last row
print(arr[:,2])#extract 3rd column
print(arr[0:3])#extract first 3 rows 
print(arr[-3:])#extract last 3 rows 
print(arr[:,0:2])#extract first 2 columns 
```

```python
print(arr[:,-2:])#extract last two columns
print(arr[0:3,1:5])#extract rows from 1 to 3 and columns from 2 to 4
print(arr[1:])#extract rows from 2nd till last 
print(arr[:,2:])#extract columns from 3 
print(arr[::2,]) #extract every 2nd row 
print(arr[:,::2]) #extract every 2nd column
```

```python
print(arr[-3:,-2:])  # extract bottom right 3x2 matrix
print(arr[0:4,0:3])  # extract top left 4x3 matrix
print(arr[::-1])     # reversing order of rows 
print(arr[:,::-1])   # reversing order of columns
print(arr[::-1,::-1])# reversing both rows & columns
print(arr[1:4,1:4])  # extracting center submatrix
print(arr[[0,-1], :])# extracting first and last rows only 
print(arr[:, [0,-1]])# extracting first and last columns only
print(arr[0:4,0:3])#extarct top left 4X3 matrix
print(arr[::-1])#reversing order of rows 
print(arr[:,::-1])#reversing order of columns
print(arr[::-1,::-1])#reversing both rows & columns
print(arr[1:4,1:4])# extracting middle 2X3 matrix 
```

```python
print(arr[0,-1])#extracting first and last rows only 
print(arr[:,0])#extracting first and last columns only
```

```python
print(arr[1::2])#extracting all odd indexed rows 
print(arr[:,::2])#extracting odd indexed columns
print(arr[4,2])#extract element at row index 4 and column index 2 
print(arr[3])#extacting entire row containing element 18
print(arr[:,3])#extracting entire column containing  value  14
```

```python
print(arr[[0,-1],[0,-1]])#extracting 4 corner elements 
print(arr[:,2:3])#extracting centre column only
print(arr[:,arr.shape[1]//2])
```

```python
print(arr[1:4,::2]) #extracting rows 1 to 4 & every second column
print(arr[:,1:4])#extracting every row but with columns index 1 to 3 
print(arr[1:4,1:4])#extract 2X2 block from centre of array
```
